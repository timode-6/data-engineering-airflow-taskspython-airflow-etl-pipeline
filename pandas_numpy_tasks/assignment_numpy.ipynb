**4.** Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B

```python
A = np.array([
    [1, 5, 3],  
    [1, 2, 5], 
    [7, 8, 9],
    [2, 4, 10],
    [3, 4, 1],
    [0, 0, 0],
    [1, 1, 1],
    [2, 3, 2]
])

B = np.array([
    [1, 2],    
    [3, 4]
])

matches = A[:, None, :, None] == B[None, :, None, :]

row_has_match_with_B_row = matches.any(axis=(2, 3))

final_mask = row_has_match_with_B_row.all(axis=1)

matching_rows = A[final_mask]

print("Matching Rows:\n", matching_rows)
```

**5.** Дана 10x3 матрица, найти строки из неравных значений (например строка [2,2,3] остается, строка [3,3,3] удаляется)

```python
matrix = np.array([
    [2, 2, 3], 
    [3, 3, 3],
    [4, 5, 6],
    [1, 1, 1],  
    [7, 8, 9],
    [5, 5, 6],
    [9, 9, 9],
    [0, 2, 4],
    [8, 8, 1],
    [6, 6, 6]
])

all_equal_mask = np.all(matrix == matrix[:, [0]], axis=1)

unequal_rows = matrix[~all_equal_mask]

print("Rows with unequal values:\n", unequal_rows)
```

* __Задача 1__: Подсчитать произведение ненулевых элементов на диагонали прямоугольной матрицы.  
 Например, для X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) ответ 3.

```python
def diag_prod_no_numpy(X):
    min_dim = min(len(X), len(X[0]))
    product = 1
    for i in range(min_dim):
        val = X[i][i]
        if val != 0:
            product *= val
    return product

X_list = [[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]
diag_prod_no_numpy(X_list)
```

```python
X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]])
diag = X.diagonal()
np.prod(diag[diag != 0])
```

```python
def is_same_multiset_no_numpy(x, y):
    if len(x) != len(y):
        return False
    return sorted(x) == sorted(y)

x_list = [1, 2, 2, 4]
y_list = [4, 2, 1, 2]
is_same_multiset_no_numpy(x_list, y_list)
```

* __Задача 3__: Найти максимальный элемент в векторе x среди элементов, перед которыми стоит ноль. 
 Например, для x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) ответ 5.

```python
def max_after_zero_no_numpy(x):
    max_val = float('-inf')
    for i in range(len(x) - 1):
        if x[i] == 0:
            max_val = max(max_val, x[i+1])
    return max_val

x_list = [6, 2, 0, 3, 0, 0, 5, 7, 0]
max_after_zero_no_numpy(x_list)
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
def rle_no_numpy(x):
    if not x:
        return ([], [])
        
    values = [x[0]]
    counts = [1]
    
    for i in range(1, len(x)):
        if x[i] == x[i-1]:
            counts[-1] += 1
        else:
            values.append(x[i])
            counts.append(1)
            
    return (values, counts)

x_list = [2, 2, 2, 3, 3, 3, 5]
rle_no_numpy(x_list)
```

```python
x  = np.array([2, 2, 2, 3, 3, 3, 5])

indices = np.nonzero(x[:-1] != x[1:])[0]

idx = np.concatenate(([-1], indices, [len(x) - 1]))

values = x[idx[1:]]
counts = np.diff(idx)

(values, counts) = (values, counts)
print((values, counts))
```

* __Задача 5__: Даны две выборки объектов - X и Y. Вычислить матрицу евклидовых расстояний между объектами. Сравните с функцией scipy.spatial.distance.cdist по скорости работы.

```python

import math
def cdist_no_numpy(X, Y):
    num_X = len(X)
    num_Y = len(Y)
    dim = len(X[0])
    
    dist_matrix = [[0.0 for _ in range(num_Y)] for _ in range(num_X)]
    
    for i in range(num_X):
        for j in range(num_Y):
            squared_sum = 0.0
            for d in range(dim):
                squared_sum += (X[i][d] - Y[j][d]) ** 2
            dist_matrix[i][j] = math.sqrt(squared_sum)
            
    return dist_matrix

X_list = [[1.0, 2.0], [3.0, 4.0]]
Y_list = [[5.0, 6.0], [1.0, 2.0], [0.0, 0.0]]
cdist_no_numpy(X_list, Y_list)
```

```python
def cdist_numpy(X, Y):
    return np.sqrt(np.sum((X[:, None, :] - Y[None, :, :]) ** 2, axis=-1))

X_np = np.array([[1.0, 2.0], [3.0, 4.0]])
Y_np = np.array([[5.0, 6.0], [1.0, 2.0], [0.0, 0.0]])
cdist_numpy(X_np, Y_np)
```
